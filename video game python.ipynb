```python
# 1.Regional Sales Distribution for Top Games

# Top 10 games by global sales
top_games = df.sort_values(
    by='Global_Sales',
    ascending=False
).head(10)

# Set game names as index
regional_sales = top_games.set_index('Name')[
    ['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']
]

print(regional_sales)


# Stacked Bar Chart
regional_sales.plot(
    kind='bar',
    stacked=True,
    figsize=(12,6)
)

plt.title("Regional Sales Distribution for Top Games")
plt.xlabel("Game Name")
plt.ylabel("Sales")

plt.xticks(rotation=90)

plt.legend(title="Region")

plt.show()
```

```python
#2.Which publisher has the highest average sales per game
df.groupby('Publisher')['Global_Sales'].mean()


avg_sales = df.groupby('Publisher')['Global_Sales'].mean().sort_values(ascending=False)

print(avg_sales.head(10))
```

```python
# Which Publisher Has the Highest Average Sales Per Game

avg_sales = df.groupby('Publisher')['Global_Sales'].mean().sort_values(ascending=False)

top_avg_sales = avg_sales.head(10)

plt.figure(figsize=(12,6))

top_avg_sales.plot(kind='bar')

plt.title("Top 10 Publishers by Average Sales Per Game")
plt.xlabel("Publisher")
plt.ylabel("Average Global Sales")

plt.xticks(rotation=90)

plt.show()
```

```python
# 3. Which Platform Had the Longest Lifespan?

platform_lifespan = df.groupby('Platform')['Year'].agg(['min', 'max'])

platform_lifespan['Lifespan'] = (
    platform_lifespan['max'] - platform_lifespan['min']
)

# Sort by lifespan
platform_lifespan = platform_lifespan.sort_values(
    by='Lifespan',
    ascending=False
)

# Top 10 longest lifespan platforms
top_platform_life = platform_lifespan.head(10)

print(top_platform_life)


# Chart
plt.figure(figsize=(12,6))

plt.bar(
    top_platform_life.index,
    top_platform_life['Lifespan']
)

plt.title("Top 10 Platforms with Longest Lifespan")
plt.xlabel("Platform")
plt.ylabel("Lifespan (Years)")

plt.xticks(rotation=90)

plt.show()
```

```python
# 4. How Did Game Releases Change Over Time?

games_per_year = df.groupby('Year')['Name'].count()

print(games_per_year)


# Line Chart
plt.figure(figsize=(12,6))

games_per_year.plot(kind='line', marker='o')

plt.title("Game Releases Over Time")
plt.xlabel("Year")
plt.ylabel("Number of Games Released")

plt.xticks(rotation=90)

plt.grid(True)

plt.show()
```

```python
# 5.Which Region Contributes Most to Global Sales?

region_sales = {
    'NA_Sales': df['NA_Sales'].sum(),
    'EU_Sales': df['EU_Sales'].sum(),
    'JP_Sales': df['JP_Sales'].sum(),
    'Other_Sales': df['Other_Sales'].sum()
}

# Convert to DataFrame
region_df = pd.DataFrame(
    list(region_sales.items()),
    columns=['Region', 'Sales']
)

print(region_df)


# Pie Chart
plt.figure(figsize=(8,8))

plt.pie(
    region_df['Sales'],
    labels=region_df['Region'],
    autopct='%1.1f%%'
)

plt.title("Regional Contribution to Global Sales")

plt.show()
```

```python
# 7.Which Genres Are Declining Over Time?

# Genre sales by year
genre_trend = df.groupby(
    ['Year', 'Genre']
)['Global_Sales'].sum().reset_index()

# Pivot table
genre_pivot = genre_trend.pivot(
    index='Year',
    columns='Genre',
    values='Global_Sales'
)

print(genre_pivot)


# Line Chart
plt.figure(figsize=(14,7))

for genre in genre_pivot.columns:
    plt.plot(
        genre_pivot.index,
        genre_pivot[genre],
        label=genre
    )

plt.title("Genre Sales Trends Over Time")
plt.xlabel("Year")
plt.ylabel("Global Sales")

plt.xticks(rotation=90)

plt.legend(bbox_to_anchor=(1.05, 1))

plt.grid(True)

plt.show()
```

```python
# 8 Publisher Genre Diversity

# Count games by Publisher and Genre
publisher_genre_count = df.groupby(
    ['Publisher', 'Genre']
).size().unstack(fill_value=0)

# Top 10 publishers by total games
top_publishers = publisher_genre_count.sum(axis=1).sort_values(
    ascending=False
).head(10).index

top_publisher_genres = publisher_genre_count.loc[top_publishers]

print(top_publisher_genres)


# Stacked Bar Chart
top_publisher_genres.plot(
    kind='bar',
    stacked=True,
    figsize=(14,7)
)

plt.title("Genre Distribution of Top Publishers")
plt.xlabel("Publisher")
plt.ylabel("Number of Games")

plt.xticks(rotation=90)

plt.legend(title="Genre", bbox_to_anchor=(1.05, 1))

plt.show()
```
